```python
import nltk
from nltk.book import *
from nltk.corpus import stopwords

# Ensure stopwords are downloaded
nltk.download('stopwords')
stop_words = set(stopwords.words('english'))

# List of the nine texts provided by 'from nltk.book import *'
texts = [text1, text2, text3, text4, text5, text6, text7, text8, text9]

print(f"{'Text':<10} | {'Tokens':<10} | {'Unique':<10} | {'Avg Length':<10}")
print("-" * 50)

for t in texts:
  # 1. Filter: Remove single-characters and stopwords
  filtered_tokens = [w for w in t if len(w) > 1 and w.lower() not in stop_words]
  # 2. Calculate Statistics
  num_tokens = len(filtered_tokens)
  num_unique = len(set(w.lower() for w in filtered_tokens))
  if num_tokens > 0:
    avg_len = sum(len(w) for w in filtered_tokens) / num_tokens
  else:
    avg_len = 0

  # 3. Output in table format
  print(f"{t.name[:8]:<10} | {num_tokens:<10} | {num_unique:<10} | {avg_len:<10.2f}")
```
